```python
import json, re
from typing import Literal, List
from pydantic import BaseModel, EmailStr, field_validator, ValidationError

Departments = Literal["IT", "HR", "Marknadsföring", "Säljare"]

class Employee(BaseModel):
    first_name: str
    last_name: str
    phone_number: str
    email: EmailStr
    department: Departments
    title: str
    salary: int

    @field_validator("first_name", "last_name", "phone_number", "department", "title", mode="before")
    @classmethod
    def clean_str(cls, v:any) -> any:
        if isinstance(v, str):
            return v.strip().replace('"', '').replace("'", "")
        return v
```
