```python
from pydantic import BaseModel
import json
class Validate_text(BaseModel):
    proposed_grade : str
    motivation : str
    improvements : str

data = json.loads(response)
result = Validate_text.model_validate(data)

pprint(result.proposed_grade)
pprint(result.motivation)
pprint(result.improvements)
```
