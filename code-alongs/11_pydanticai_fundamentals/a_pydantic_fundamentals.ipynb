## Pydantic fundamentals 

```python
from pydantic_ai import Agent
from dotenv import load_dotenv

load_dotenv

agent = Agent(model="google-gla:gemini-2.5-flash")
```

```python
result = await agent.run("Give me a it person working it in sweden")
print(result.output)
```

```python
from pydantic import BaseModel, Field

class EmployeeModel(BaseModel):
    name: str = Field(description="Name of employye")
    Age: int = Field(description="Age of employye", gt = 17, lt =70)
    Location: str = Field(description="location of employye")
    Role: str = Field(description="Role of employye")
    Company: str = Field(description="Company where employye is employed")

result = await agent.run("Give me an it person working it in sweden", output_type=EmployeeModel)
print(result.output)
```
