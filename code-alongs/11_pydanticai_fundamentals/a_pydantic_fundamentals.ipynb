## Pydantic fundamentals 

```python
from pydantic_ai import Agent
from dotenv import load_dotenv

load_dotenv

agent = Agent(model="google-gla:gemini-2.5-flash")
```

```python
result = await agent.run("Say hello in one short sentence.")
print(result.output)
```
