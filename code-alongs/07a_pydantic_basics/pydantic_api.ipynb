## Deserialization - turn json object in to Pydantic model

- data has a key called results
- inside of results it has value of list of dictionaries
- another key called count
- choose thhe fields tou want to include

```python
from pydantic import BaseModel

class PokemonListResponse(BaseModel):
    results: list[dict]
    count: int

pokemons = PokemonListResponse.model_validate(data)
pokemons
```

```python
pokemons.count
```

```python
pokemons.results
```

```python
from pydantic import computed_field

class PokemonListResponse(BaseModel):
    results: list[dict]
    count: int

    @computed_field
    @property
    def number_strored_pokemon(self) -> int:
        return len(self.results)
    

pokemons = PokemonListResponse.model_validate(data)
pokemons.number_strored_pokemon
```

## serialize into json

```python
with open("pokemons.jason", "w") as file:
    file.write(pokemons.model_dump_json(indent=3))
```
